```python
import numpy as np

# Environment setup
states = list(range(8))  # Positions 0 to 4 in a row
actions = ['pick', 'skip']  # Possible actions
rewards = {0: 1, 1: -1, 2: 1, 3: -1, 4: 1, 5: 1, 6:-1, 7:-1}  # Ripe (+1) or unripe (-1) apples
gamma = 0.9  # Discount factor
theta = 0.01  # Convergence threshold

# Initialize random policy and value function
policy = {s: np.random.choice(actions) for s in states}
values = {s: 0.0 for s in states}

def get_next_state(state, action):
    """Determine next state based on action."""
    return min(state + 1, len(states) - 1)  # Move to next position or stay at end

def policy_evaluation(policy, values):
    """Evaluate the value function for a given policy."""
    while True:
        delta = 0
        for s in states:
            v = values[s]
            a = policy[s]
            next_state = get_next_state(s, a)
            reward = rewards[s] if a == 'pick' else 0
            values[s] = reward + gamma * values[next_state]
            delta = max(delta, abs(v - values[s]))
        if delta < theta:
            break
    return values

def policy_improvement(policy, values):
    """Improve the policy based on the value function."""
    policy_stable = True
    for s in states:
        old_action = policy[s]
        # Compute value for each action
        action_values = {}
        for a in actions:
            next_state = get_next_state(s, a)
            reward = rewards[s] if a == 'pick' else 0
            action_values[a] = reward + gamma * values[next_state]
        # Choose action with highest value
        policy[s] = max(action_values, key=action_values.get)
        if old_action != policy[s]:
            policy_stable = False
    return policy, policy_stable

def visualize_row(policy, values):
    print("Policy: 'pick' = Pick apple, 'skip' = Move to next position")

    """Visualize the apple row environment once with policy and values."""
    apple_states = {s: '🍎' if rewards[s] == 1 else '🥔' for s in states}
    
    # Show the apple row once
    print("\nApple Row (🍎 = Ripe, 🥔 = Unripe):")
    row = ['-'] * len(states)
    for s in states:
        row[s] = apple_states[s]
    print("  " + " ".join(row))
    print("  " + " ".join(str(s) for s in states))
    print(f"Current Robot Policy: {policy}")
    
    # # Detailed table
    # print("\nDetailed State Information:")
    # print("Position | Apple | Policy | Value")
    # print("-" * 35)
    # for s in states:
    #     print(f"{s:8} | {apple_states[s]:5} | {policy[s]:6} | {values[s]:.3f}")
    
    # Legend


# Visualize the initial state
visualize_row(policy, values)

# Policy iteration
print("----Initial Policy:", policy)
for i in range(10):  # Max iterations
    # Policy evaluation
    values = policy_evaluation(policy, values.copy())
    print(f"Iteration {i+1} - Values:", values)
    # Policy improvement
    policy, stable = policy_improvement(policy, values.copy())
    print(f"Iteration {i+1} - Policy:", policy)
    if stable:
        print("Policy converged!")
        break

# Final output with single visualization
print("\nOptimal Policy:", policy)
print("Optimal Values:", values)
```
