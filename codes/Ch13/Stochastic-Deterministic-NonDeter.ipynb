```python
import numpy as np

class GridWorld:
    def __init__(self, size=5):
        self.size = size
        self.actions = ['Up', 'Down', 'Left', 'Right']
        self.action_map = {'Up': (-1, 0), 'Down': (1, 0), 'Left': (0, -1), 'Right': (0, 1)}
        self.state = (0, 0)  # Start at top-left corner

    def reset(self):
        self.state = (0, 0)
        return self.state

    def deterministic_step(self, action):
        """Deterministic transition: Move exactly in the action direction."""
        move = self.action_map[action]
        new_state = (self.state[0] + move[0], self.state[1] + move[1])
        # Check boundaries
        new_state = (
            max(0, min(new_state[0], self.size - 1)),
            max(0, min(new_state[1], self.size - 1))
        )
        self.state = new_state
        return self.state

    def stochastic_step(self, action):
        """Stochastic transition: 80% intended direction, 10% each to adjacent directions."""
        probs = {'Up': 0.1, 'Down': 0.1, 'Left': 0.1, 'Right': 0.1}
        probs[action] = 0.8  # 80% chance for intended action
        # Remove opposite direction to normalize probabilities
        opposites = {'Up': 'Down', 'Down': 'Up', 'Left': 'Right', 'Right': 'Left'}
        del probs[opposites[action]]
        probs = {k: v / sum(probs.values()) for k, v in probs.items()}  # Normalize
        # Choose actual action based on probabilities
        actual_action = np.random.choice(list(probs.keys()), p=list(probs.values()))
        return self.deterministic_step(actual_action)  # Apply deterministic move

    def nondeterministic_step(self, action):
        """Nondeterministic transition: Equal probability to all neighboring states."""
        neighbors = []
        # Possible moves: Up, Down, Left, Right, Stay
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1), (0, 0)]
        for move in moves:
            new_state = (self.state[0] + move[0], self.state[1] + move[1])
            # Check if within bounds
            if 0 <= new_state[0] < self.size and 0 <= new_state[1] < self.size:
                neighbors.append(new_state)
        # Choose a random neighbor
        self.state = neighbors[np.random.randint(len(neighbors))]
        return self.state

def simulate_process(grid, process_type, steps=5):
    """Simulate the specified process type for a number of steps."""
    print(f"\nSimulating {process_type} Process:")
    print(f"Initial state: {grid.reset()}")
    for step in range(steps):
        # Choose a random action for simplicity
        action = np.random.choice(grid.actions)
        if process_type == "Deterministic":
            new_state = grid.deterministic_step(action)
        elif process_type == "Stochastic":
            new_state = grid.stochastic_step(action)
        elif process_type == "Nondeterministic":
            new_state = grid.nondeterministic_step(action)
        print(f"Step {step + 1}: Action = {action}, New state = {new_state}")
```

```python
def main():
    # Set random seed for reproducibility
    np.random.seed(42)
    # Create gridworld
    grid = GridWorld(size=5)
    # Simulate each process
    simulate_process(grid, "Deterministic")
    simulate_process(grid, "Stochastic")
    simulate_process(grid, "Nondeterministic")

if __name__ == "__main__":
    main()
```
