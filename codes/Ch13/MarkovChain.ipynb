```python
import numpy as np
import random

def simulate_markov_chain(start_state, transition_matrix, num_steps):
    current_state = start_state
    chain_sequence = [current_state]

    for _ in range(num_steps - 1):
        # Get the probabilities for the current state's transitions
        possible_next_states = list(transition_matrix[current_state].keys())
        probabilities = list(transition_matrix[current_state].values())

        # Choose the next state randomly based on probabilities
        next_state = random.choices(possible_next_states, weights=probabilities)[0]
        chain_sequence.append(next_state)
        current_state = next_state

    return chain_sequence
```

```python

# Example: Weather states (Sunny, Cloudy, Rainy)
states = ['Sunny', 'Cloudy', 'Rainy']

# Transition matrix: rows represent current state, columns represent next state
# Each row must sum to 1
transition_matrix = {
    'Sunny': {'Sunny': 0.7, 'Cloudy': 0.2, 'Rainy': 0.1},
    'Cloudy': {'Sunny': 0.3, 'Cloudy': 0.5, 'Rainy': 0.2},
    'Rainy': {'Sunny': 0.1, 'Cloudy': 0.3, 'Rainy': 0.6}
}
# Example usage:
start_state = 'Sunny'
num_steps = 10

simulated_chain = simulate_markov_chain(start_state, transition_matrix, num_steps)
print(f"Simulated Markov Chain: {simulated_chain}")
```
